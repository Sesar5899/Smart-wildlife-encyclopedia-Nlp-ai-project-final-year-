# Ecosystem and protected-area Burmese text-quality baseline audit

This notebook preserves the findings from the read-only baseline audit. The 13 high-confidence ecosystem findings documented below were corrected on 2026-09-22; the protected-area findings were not changed by that remediation.

**Scope:** 62 ecosystem records (27 `_mm` fields; 1,312 non-empty cells) and 39 protected-area records (21 `_mm` fields; 610 non-empty cells).


## Summary

- High-confidence visible typo or repetition findings: 13 ecosystems and 5 protected areas.
- Specific words using non-standard `ဥ်` where standard spelling uses `ဉ်`: all 62 ecosystems and 23 protected areas.
- Review-only wording variants: `အိမ်ယာ/အိမ်ရာ` and `ဒဟက်/ဒဟတ်`. These are not counted as confirmed errors.
- Canonically equivalent Unicode forms such as `ဦး/ဦး` are excluded from spelling-error counts.


```python
import json, re, unicodedata
from pathlib import Path

ecosystems = json.loads(Path('../data/ecosystems.json').read_text(encoding='utf-8-sig'))
protected_areas = json.loads(Path('../data/protected_areas.json').read_text(encoding='utf-8-sig'))

def scope(records):
    fields = sorted({key for row in records for key in row if key.endswith('_mm')})
    nonempty = sum(bool(str(row.get(field, '')).strip()) for row in records for field in fields)
    return len(records), len(fields), nonempty

scope(ecosystems), scope(protected_areas)
```

## Non-standard `ဥ်` usage

All 62 ecosystem records contain at least one specific word using `ဥ်` where standard spelling uses `ဉ်`. The protected-area records affected are listed below.

`alaungdaw_kathapa_national_park_protected`, `ashi_myin_anawt_myin_taung_protected`, `eimawbon_national_park_protected`, `gulf_of_mottama_protected`, `hkakaborazi_national_park_protected`, `hlawga_wildlife_park_protected`, `hponkanrazi_wildlife_sanctuary_protected`, `htamanthi_wildlife_sanctuary_protected`, `indawgyi_wildlife_sanctuary_protected`, `inlay_lake_wildlife_sanctuary_protected`, `kyeeyeyan_taung_national_park_protected`, `loimwe_protected_area_protected`, `mainmahla_kyun_wildlife_sanctuary_protected`, `me_a_pi_taung_protected`, `minsontaung_wildlife_sanctuary_protected`, `natmataung_national_park_protected`, `panlaung_pyadalin_cave_wildlife_sanctuary_protected`, `parsar_protected_area_protected`, `rakhine_yoma_elephant_range_protected`, `sar_munm_national_park_protected`, `shwesettaw_wildlife_sanctuary_protected`, `taninthayi_national_park_protected`, `thamihla_kyun_wildlife_sanctuary_protected`.


```python
def scan_u1(records):
    findings = []
    for row in records:
        fields = [key for key, value in row.items() if key.endswith('_mm') and 'ဥ်' in str(value)]
        if fields:
            findings.append({'record': row.get('id') or row.get('name'), 'name_mm': row.get('name_mm'), 'fields': fields})
    return findings

u1_ecosystems = scan_u1(ecosystems)
u1_protected_areas = scan_u1(protected_areas)
len(u1_ecosystems), len(u1_protected_areas)
```

`အိမ်ယာ/အိမ်ရာ` occurs inconsistently in 16 ecosystem records and 4 protected-area records. `ဒဟက်/ဒဟတ်` occurs inconsistently in 3 ecosystem records. These are flagged for editorial review, not declared errors, because source terminology may be intentional.
