# Standard Burmese spelling field audit

Read-only audit of the 58 requested animal IDs. The database is opened with SQLite `mode=ro`; no data is changed.


```python
import re, sqlite3

ids = [1,2,3,4,5,6,9,10,12,13,14,16,17,18,19,20,21,24,25,26,28,30,31,32,33,34,35,36,37,39,41,42,43,44,45,46,48,53,55,56,57,59,60,61,62,67,68,69,70,71,74,76,77,78,79,80,81,106]
pattern = re.compile(r'(အစီအစဥ်|စီစဥ်|မျိုးသုဥ်း|အလျဥ်း|လျဥ်း|ယဥ်း|ကျဥ်း|ပျဥ်း|ဥယျာဥ်|ယာဥ်|စဥ်|ဦး|အဆောက်အဦး)')
db = sqlite3.connect('file:../wildlife.db?mode=ro', uri=True)
db.row_factory = sqlite3.Row
columns = [row[1] for row in db.execute('PRAGMA table_info(animals)')]
fields = list(dict.fromkeys(['name_my'] + [c for c in columns if c.endswith('_my')] + ['conservation_status','population_trend','myanmar_population','global_population','other_name']))
results = []
for animal_id in ids:
    animal = db.execute('SELECT * FROM animals WHERE id=?', (animal_id,)).fetchone()
    hits = [field for field in fields if isinstance(animal[field], str) and pattern.search(animal[field])]
    results.append((animal_id, animal['name_my'], hits))
results
```

## Result summary

All 58 requested IDs currently match at least one checked spelling pattern, covering 122 animal-field combinations.
